# STEP 1: Setup the environment for Data Analysis 
* Initialize python/pandas
* Import the "consolidated data.xlsx" Excel worksheet and **print a simplified version to browser**

```python
import pandas as pd
pd.set_option("display.max_rows", 500)
import openpyxl
import webbrowser
print("pandas version used is ", pd.__version__)

# Read Excel file with DataFrame type annotation
df: pd.DataFrame = pd.read_excel("consolidated data.xlsx", sheet_name="HF R37-16 Consolidated Data")

# Select specific columns with DataFrame type annotation
newdf: pd.DataFrame = df[["Enumerator", "Form Number", "District", "Category of Place"]].copy()

# Clean up extra spaces in text columns
newdf["District"] = newdf["District"].astype(str).str.strip()

# Sort by District and Category of Place
sorted_df: pd.DataFrame = newdf.sort_values(by=["District", "Category of Place"])

# Groupby District and Category of Place
sorted_grouped_df = newdf.groupby(["District", "Category of Place"])

# Export sorted_df to HTML
sorted_df.to_html("sorted.html")
webbrowser.open("sorted.html")
```
