```python
import sys
import pandas as pd

# 1. Print Pandas and Python versions correctly
print("working with pandas version:", pd.__version__)
print("working with python version:", sys.version.split()[0])

# 2. Define the list
kids = [100, 151, 200]

# 3. Print the data type of each individual item in the list
for kid in kids:
    print(f"Item: {kid} -> Type: {type(kid)}")
```

```python
# Convert the array object to a Pandas Series. 
s= pd.Series(kids)
# print (f"type of \"s\" object  is: \t {type(s)}")
print (f"printing standard series object\n {s}\n")

# Create a pandas Series from the array object with a custom index. 
s= pd.Series(kids, index=["First","Second","Third"])
print (f"printing series created with custom index \n {s}")

# Accessing individual elements in series  with iloc[] property 
print("\n accessing individual items with iloc[] property which ONLY WORKS WITH INT BASED INDEXES")
for i in range(0,len(s)):
    print(f"the item at location {i} is {s.iloc[i]}") 


# Accessing individual elements in series  with loc[] property 
print("\n accessing individual items with loc[] property which also works with custom indexes")
for idx in s.index:
    print(f"the item at location {idx} is {s.loc[idx]}") # loc() can also work with custom indexes 
```

```python
# conditionally print series members 
print(s[s>150])
```

```python
daysofWeek = {
    "Sunday": 0,
    "Monday": 1,
    "Tuesday": 2,
    "Wednesday": 3,
    "Thursday": 4,
    "Friday": 5,
    "Saturday": 6
}

# Convert dictionary to Pandas Series
s_days = pd.Series(daysofWeek, name="Days_of_Week")

print ("FIRST WEEK\n")
print(s_days)

# increment by 7 days ro represente next week 
for i in range (0,len(s_days)):
    s_days.iloc[i] +=7
print("\n Next week \n")


print(s_days.loc["Wednesday"])
print(s_days)
```
